```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
img = cv2.imread('images (1).jpeg')
```

```python
plt.imshow(img[:,:,::-1])
```

```python
np.max(img)//2
```

```python
canny = cv2.Canny(img,127,127)
```

```python
plt.imshow(canny,cmap='grey')
```

```python
blur_img= cv2.blur(img,(3,3))
```

```python
plt.imshow(blur_img[:,:,::-1])
```

```python
medium = np.median(blur_img)
```

```python
low = int(max(0,0.7*medium))
high=int(max(255,1/3*medium))
print(low,medium,high)
```

```python
canny_img = cv2.Canny(blur_img,105,255)
```

```python
plt.imshow(canny_img,cmap='grey')
```
